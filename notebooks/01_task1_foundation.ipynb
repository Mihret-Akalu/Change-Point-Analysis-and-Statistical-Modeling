## 2. Assumptions and Limitations

### 2.1 Key Assumptions:

* **Stationarity**: Log returns approximate stationarity—useful for change point frameworks.
* **Abruptness**: Bayesian change point models assume regime shifts are abrupt rather than gradual.
* **Coincidence**: Events coincide with structural shifts but are not directly proven causal drivers.

### 2.2 Limitations:

* **Causality**: Identifying a temporal overlap between a detected change point and an event is not evidence of causation.
* **Complexity**: External variables (macroeconomic, policy changes) are not explicitly modeled.
* **Lag**: Some events have lagged or delayed effects that may not align exactly with price changes.

---

## 3. Change Point Model Overview

We plan to use a Bayesian model where the change point index  is given a discrete uniform prior over all days. Regime parameters (means, and possibly variances) are estimated for segments before and after this point. In **PyMC**, the `pm.math.switch()` function will define a step in the mean depending on whether the index is before or after the change point.
